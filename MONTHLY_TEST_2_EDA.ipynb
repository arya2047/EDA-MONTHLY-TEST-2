DATA UNDERSTANDING & CLEANING

```python
df=pd.read_csv('/content/DA_Month2_EDA_Machine_Test_Dataset.csv')
```

```python
display(df.shape)
display(df.columns)
display(df.dtypes)
display(df.head())
```

```python
df['Region']=(df['Region'].str.strip().str.title())
```

```python
df['Customer_Segment']=(df['Customer_Segment'].str.strip().str.title())
```

```python
df['Category']=(df['Category'].str.strip().str.title())
```

```python
df['Product']=(df['Product'].str.strip().str.title())
```

```python
df.isnull().sum()
```

```python
df['Region']=df['Region'].fillna(df['Region'].mode()[0])
df['Region'].isnull().sum()
```

```python
df['Customer_Segment']=df['Customer_Segment'].fillna(df['Customer_Segment'].mode()[0])
df['Customer_Segment'].isnull().sum()
```

```python
df['Payment_Method']=df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])
df['Payment_Method'].isnull().sum()
```

```python
df.duplicated()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated()
```

```python
df
```

```python
Q1=df['Quantity'].quantile(0.25)
Q3=df['Quantity'].quantile(0.75)

IQR=Q3-Q1

upper_limit=Q3-1.5*IQR
lower_limit=Q3+1.5*IQR

df_clean=df[(df['Quantity']>=lower_limit)&
(df['Quantity']<=upper_limit)]

print(df_clean)
```

```python
Q1=df['Discount'].quantile(0.25)
Q3=df['Discount'].quantile(0.75)

IQR=Q3-Q1

upper_limit=Q3-1.5*IQR
lower_limit=Q3+1.5*IQR

df_clean=df[(df['Discount']>=lower_limit)&
(df['Discount']<=upper_limit)]

print(df_clean)
```

```python
df['Region']=(df['Region'].str.strip().str.title())
```

```python
orders = df.groupby('Order_ID')['Sales'].sum()
plt.figure(figsize=(10, 5))
plt.plot(orders.index, orders.values)
plt.xlabel('Order ID')
plt.ylabel('Total Sales')
plt.title('Order ID vs Total Sales')
plt.xticks([])
plt.show()
```

```python
plt.bar(df['Region'],df['Sales'])
plt.xlabel('Region')
plt.ylabel('Sales')
plt.title('Sales across available regions/locations')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=df.groupby("Quantity")['Sales'].sum()
plt.plot(data.index,data.values)
plt.xlabel('Quantity')
plt.ylabel('Total Sales')
plt.title('Quantity vs Total Sales')
plt.show()
```

```python
cols=['Quantity','Discount','Sales','Unit_Price']

for col in cols:

  sns.histplot(data=df,color='blue')
plt.xlabel('Columns')
plt.ylabel('Frequency')
plt.title('Distributions')
plt.xlim(0,1000)
plt.ylim(0,1000)
plt.show()
```

```python
cols=['Quantity','Unit_Price','Discount','Sales']
correlation=df[cols].corr()
sns.heatmap(correlation,annot=True,cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.title('Category vs Total Sales')
plt.show()
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'])
df.groupby(df['Order_Date'].dt.month)['Sales'].sum().plot(kind='bar')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.title('Monthvs Total Sales')
plt.show()
```
